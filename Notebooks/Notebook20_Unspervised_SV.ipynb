## Unsupervised Learning

In our previous notebooks, we explored supervised learning, where our goal was to predict an outcome variable using a set of predictors. For example, we used logistic regression to predict whether breast tumors were benign or malignant.

In this final notebook, we will introduce unsupervised learning, where the goal is not to predict an outcome but instead to identify patterns and relationships within a dataset.

We will focus on two methods:

- Principal Component Analysis (PCA): Reducing a large number of variables into a smaller number of components while retaining as much variation as possible.
- K-Means Clustering: Identifying groups of observations that share similar characteristics without knowing their group memberships beforehand.

For this analysis, we will use a dataset containing chemical measurements from 178 wine samples belonging to three different wine classes. Our goal will be to examine whether these measurements reveal meaningful patterns and groupings.
