**Applicant Information**

* Gender — Gender recorded for the applicant (Male/Female).
* Married — Whether the applicant is married (Yes/No).
* Dependents — Number of people financially dependent on the applicant.
* Education — Educational background of the applicant (Graduate/Not Graduate).
* Self_Employed — Whether the applicant is self-employed (Yes/No).

**Financial Information**

* ApplicantIncome — Income earned by the primary loan applicant.
* CoapplicantIncome — Income earned by an additional applicant, if applicable.
* LoanAmount — Amount of money requested for the loan, recorded in thousands.
* Loan_Amount_Term — Length of time allowed to repay the loan, measured in months.
* Credit_History — Indicates whether the applicant's credit history meets the lender's guidelines (1 = meets guidelines, 0 = does not meet guidelines).

**Property and Loan Decision**

* Property_Area — Type of area where the property is located (Urban, Semiurban, or Rural).
* Loan_Status — Whether the loan application was approved (Y = Approved, N = Not Approved).
