Here is some additional information on the columns/variables in this dataset:

**General Vehicle Information**

* symboling — Insurance risk rating assigned to a vehicle, ranging from -3 (lower risk) to +3 (higher risk).
* normalized-losses — Average insurance loss associated with the vehicle, adjusted for its vehicle classification.
* make — Manufacturer or brand of the vehicle.
* fuel-type — Type of fuel used by the vehicle (gas or diesel).
* aspiration — Whether the engine is naturally aspirated (std) or turbocharged (turbo).
* num-of-doors — Number of doors on the vehicle (two or four).
* body-style — Style of the vehicle's body (sedan, hatchback, wagon, etc.).
* drive-wheels — Which wheels receive power from the engine (front, rear, or four-wheel drive).
* engine-location — Location of the vehicle's engine (front or rear).
Vehicle Size and Weight
* wheel-base — Distance between the front and rear axles, measured in inches.
* length — Overall length of the vehicle, measured in inches.
* width — Overall width of the vehicle, measured in inches.
* height — Overall height of the vehicle, measured in inches.
* curb-weight — Weight of the vehicle without passengers or cargo, measured in pounds.

**Engine Specifications**

* engine-type — Design or configuration of the vehicle's engine.
* num-of-cylinders — Number of cylinders in the engine.
* engine-size — Total engine displacement, measured in cubic inches.
* fuel-system — Type of system used to deliver fuel to the engine.
* bore — Diameter of the engine's cylinders, measured in inches.
* stroke — Distance the piston travels inside a cylinder, measured in inches.
* compression-ratio — Ratio describing how much the air-fuel mixture is compressed inside the engine.
* horsepower — Amount of power produced by the engine.
* peak-rpm — Engine speed, in revolutions per minute, at which maximum horsepower is produced.

**Fuel Efficiency and Price**

* city-mpg — Fuel efficiency when driving in city conditions, measured in miles per gallon.
* highway-mpg — Fuel efficiency when driving on highways, measured in miles per gallon.
* price — Price of the automobile in US dollars at the time of the dataset.
